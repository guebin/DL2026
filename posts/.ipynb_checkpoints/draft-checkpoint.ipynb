`# 예제1` -- 아래의 이차함수

$$f(x)=(x-2)^2$$

를 최소화하는 $x$값을 구하라. (편의상 답은 $x_{sol}$이라고 하자)

`(풀이1)`

완전제곱식의 성질을 생각하면 답은 2이다. 따라서 $x_{sol}=2$.

`(풀이2)`

$f''(x)>0$ 이므로 이 함수는 아래로 볼록한 이차함수이다. 따라서 $f'(x)=2(x-2)=0$ 를 만족하는 $x$가 최소값이다. 따라서 $x_{sol}=2$. 

`(풀이3)` 

> *가정: $f(x)$는 아래로 볼록한 이차함수라고 가정하자.*

`-` 아래와 같은 알고리즘을 생각하자. 

1. 아무 값이나 $x_{sol}$이라 주장한다. 
2. 1에서 선택한 $x_{sol}$값에서 좌우로 조금씩 이동해보고 더 $f(x)$를 작게 만드는 값으로 update한다.  
3. 업데이트된 $x_{sol}'$을 다시 $x_{sol}$로 해석하고 1-2를 반복한다.

````python
def f(x):
    return (x-2)**2
````

`# 경사하강법 아이디어 (1차원)`

1. 임의의 점을 찍는다.
2. 그 점에서 순간변화율을 구한다. <-- 미분
3. 순간변화율의 부호를 살펴보고 부호와 반대방향으로 움직인다. 이때 기울기의 절대값 크기와 비례하여 보폭(=움직이는 정도)을 각각 조절한다. 즉 아래의 수식에 따라 업데이트 한다.
$$ x_{\text{sol}} \leftarrow x_{\text{sol}} - \alpha \times f'(x_{\text{sol}})$$

`#`

````python
alpha = 0.3
````

````python
x_sol = 2.23
x_sol
````

````python
slope = 2*(x_sol-2)
x_sol = x_sol - alpha * slope 
x_sol
````

`-` 시각화 

````python
x = torch.linspace(1.7,2.3,100)
plt.plot(x,f(x))
#---#
# 초기값
x_sol = 2.23
plt.plot(x_sol,f(x_sol),'o')
# 1회 update
slope = 2*(x_sol-2)
x_sol = x_sol - alpha * slope
plt.plot(x_sol,f(x_sol),'o')
# 2회 update
slope = 2*(x_sol-2)
x_sol = x_sol - alpha * slope
plt.plot(x_sol,f(x_sol),'o')
# 3회 update
slope = 2*(x_sol-2)
x_sol = x_sol - alpha * slope
plt.plot(x_sol,f(x_sol),'o')
````

`-` 반복연산 

````python
x_sol = 2.23
for dummy in range(100):
    slope = 2*(x_sol-2)
    x_sol = x_sol - alpha*slope
x_sol
````

`-` 애니메이션 

````python
x_sol_trace = [2.23]
for i in range(100):
    slope = 2*(x_sol_trace[i]-2)
    x_sol_trace.append(x_sol_trace[i] - alpha*slope)
x_sol_trace = torch.tensor(x_sol_trace)
y_trace = f(x_sol_trace)
fig = plt.figure()
ax = fig.gca()
ax.plot(x,f(x))
point, = ax.plot(x_sol_trace[0],y_trace[0],'o',color="C1")
def func(frame):
    point.set_xdata(x_sol_trace[:frame])
    point.set_ydata(y_trace[:frame])
    return point, 
ani = matplotlib.animation.FuncAnimation(
    fig, func, frames=10
)
display(IPython.display.HTML(ani.to_jshtml()))
````

````python
plt.rcParams["animation.html"] = "jshtml" 
````

````python
def show_plot(alpha=0.001):
    x_sol_trace = [2.23]
    for i in range(100):
        slope = 2*(x_sol_trace[i]-2)
        x_sol_trace.append(x_sol_trace[i] - alpha*slope)
    x_sol_trace = torch.tensor(x_sol_trace)
    y_trace = f(x_sol_trace)

    fig = plt.figure()
    ax = fig.gca()
    ax.plot(x, f(x))
    point, = ax.plot(x_sol_trace[0], y_trace[0], 'o', color="C1")

    def func(frame):
        point.set_xdata(x_sol_trace[:frame])
        point.set_ydata(y_trace[:frame])
        return point,

    ani = matplotlib.animation.FuncAnimation(fig, func, frames=10)
    plt.close()
    return ani
````

````python
ani = show_plot(alpha=0.01)
ani
````

`# 우리의 무기` -- 경사하강법 

함수 $f(x)$가 아래로 볼록한 모양이라고 하자. $f(x)$를 최소로 만드는 $x$값을 찾기 위해서는 아래의 알고리즘을 사용하면 된다. 

1. 최초의 $x$값을 아무값이나 선택한다. 
2. 아래의 수식을 이용하여 update 한다. 
$$x \leftarrow x-\alpha \times f'(x)$$
3. 1-2를 무한반복한다. (실무에서는 적당한 횟수에서 마무리함)

주의점: $\alpha$를 잘 선택해야한다.

### `풀이4` -- 저 $f'(x)$ 구할줄 모르는데요?

`-` 아래를 미분할줄 모를수도 있음

$$f(x)=\frac{1}{4}(x-2)^2$$

````python
def f(x):
    return (1/4)*(x-2)**2
````

`-` 그렇지만 미분값을 근사계산 할 수는 있음. 

$$f'(x) \approx \frac{f(x+h)-f(x)}{h}$$

````python
x = 6
(f(x+0.01)-f(x))/0.01
````

`-` 그러면 별거없는데??

````python
# 기존코드
def ff(x):
    return 1/2*(x-2)
x = 6
alpha = 1
for dummy in range(10):
    x = x - alpha*ff(x)
print(x)
# 수정코드
x = 6
alpha = 1
for dummy in range(10):
    x = x - alpha*((f(x+0.01)-f(x))/0.01)
print(x)
````

### `풀이5` -- 파이토치 활용

## C. 박혜원씨의 추축

`-` 이거 데이터를 보니까 꼭 $y_i \approx \beta_0 + \beta_1 x_i$인 것 같지않어? 

````python
plt.plot(x,y,'o',label=r"observed data: $(x_i,y_i)$")
plt.legend()
````

`-` 박혜원씨의 의심: 이 세상 혹시 시뮬레이션 아니야???

- 내가 관측한 sales는 사실 $\text{sales}_i = \beta_0 + \beta_1 \text{temp}_i + \epsilon_i, \epsilon_i \overset{i.i.d.}{\sim} N(0,??)$ 에서 얻어진것 같아.. 
- 증거? 그냥 느낌이 그래.. (그림을 봐라, 딱 보면 몰라??) <--- 검증이 필요함 (오차항분석)

`-` 박혜원씨의 생각: 앞으로 이 세상은 시뮬리에션이라고 가정하자. 

- 본질적으로 `sales`와 `temp` 사이에는 직선의 관계가 있어야한다. ($x_i$, $y_i$는 선형성을 가짐) 
- 그런데 왜 그림을 그려보면 딱 직선이 보이지 않냐? 누군가가 각 관측치에 장난질을 쳐서^[좀 더 구체적을 말하면 각 관측치에 독립적으로 장난질을 치는것 같고 그리고 동일한 장난질을 치는듯 하다] 본질을 흐리고 있기 때문이다. (=`sales`를 흔들리게 하고 있기 때문이다.)
- 나는 장난질에 흔들리지 않고 본질을 꿰뚫어 보겠다. 그래서 커피판매량을 잘 예측해서 부자가 되겠다. 

`-` 본질을 꿰뚫어봄 = $(x_i,y_i)$를 "적당히 잘 관통하는" 하나의 추세선을 그리는 것 = $y_i \approx \beta_0 + \beta_1x_i$를 최~~~대한 비슷하게 만드는 $\beta_0$와 $\beta_1$을 잘 고름. 

`-` 앞으로 편의상 아래와 같이 노테이션을 통일하자. (왼쪽은 통계학과 스타일, 오른쪽은 딥러닝 스타일) 

$$y_i = \beta_0 + \beta_1 x_i +\epsilon_i = w_0 + w_1 x_i + \epsilon_i$$

## B. 2단계 -- update 

`-` ’적당한 정도’를 판단하기 위한 장치: loss function 도입!

$$\begin{align*}
loss=& \sum_{i=1}^{n}(y_i-\hat{y}_i)^2=\sum_{i=1}^{n}(y_i-(\hat{w}_0+\hat{w}_1x_i))^2\\
=&({\bf y}-{\bf\hat{y}})^\top({\bf y}-{\bf\hat{y}})=({\bf y}-{\bf X}{\bf \hat{W}})^\top({\bf y}-{\bf X}{\bf \hat{W}})
\end{align*}$$

`-` loss 함수의 특징: 위 그림의 주황색 점선이 ‘적당할 수록’ loss값이 작다.

````python
plt.plot(x,y,'o')
plt.plot(x,yhat)
````

````python
loss = torch.sum((y-yhat)**2)
loss
````

- 궁극적으로는 아예 모든 조합 $(\hat{w}_0,\hat{w}_1)$에 대하여 가장 작은 loss를 찾으면 좋겠다. 

`-` 문제의 치환: 생각해보니까 우리의 문제는 아래와 같이 수학적으로 단순화 되었다.

-   가장 적당한 주황색 선을 찾자 $\to$ $loss(\hat{w}_0,\hat{w}_1)$를 최소로하는 $(\hat{w}_0,\hat{w}_1)$의 값을 찾자.

`-` 수정된 목표: $loss(\hat{w}_0,\hat{w}_1)$를 최소로 하는 $(\hat{w}_0,\hat{w}_1)$을 구하라.

-   단순한 수학문제가 되었다. 이것은 마치 $f(x,y)$를 최소화하는
    $(x,y)$를 찾으라는 것임.

-   최적화를 하는 방법? 경사하강법


`# 경사하강법 아이디어 (1차원)`

1.  임의의 점을 찍는다.
2.  그 점에서 순간기울기를 구한다. (접선) \<– 미분
3.  순간기울기(=미분계수)의 부호를 살펴보고 부호와 반대방향으로 움직인다.

> 팁: 기울기의 절대값 크기와 비례하여 보폭(=움직이는 정도)을 조절한다.
> $\to$ $\alpha$를 도입

> 최종수식:
> $\hat{w} \leftarrow \hat{w} - \alpha \times \frac{\partial}{\partial w}loss(w)$

`#`

`# 경사하강법 아이디어 (2차원)`

![](https://guebin.github.io/DL2024/posts/02wk-1_files/figure-html/cell-37-output-1.png)

1.  임의의 점을 찍는다.
2.  그 점에서 순간기울기를 구한다. (접평면) \<– 편미분
3.  순간기울기(=미분계수)의 부호를 살펴보고 부호와 반대방향으로 각각 움직인다.

> 팁: 여기서도 기울기의 절대값 크기와 비례하여 보폭(=움직이는 정도)을 각각 조절한다. $\to$ $\alpha$를 도입.


`#`

`-` 경사하강법 = **loss를 줄이도록 ${\bf \hat{W}}$를 개선하는 방법**

-   업데이트 공식: 수정값 = 원래값 - $\alpha$ $\times$ 기울어진크기(=미분계수)
-   여기에서 $\alpha$는 전체적인 보폭의 크기를 결정한다. 즉 $\alpha$값이 클수록 한번의 update에 움직이는 양이 크다.

`-` loss는 $\hat{\bf W} =\begin{bmatrix} \hat{w}_0 \\ \hat{w}_1 \end{bmatrix}$ 에 따라서 값이 바뀌는 함수로 해석가능하고 구체적인 형태는 아래와 같음.

- $loss(\hat{w}_0,\hat{w}_1) =\sum_{i=1}^{n}(y_i-(\hat{w}_0+\hat{w}_1x_i))^2$
- $loss(\hat{\bf W})=({\bf y}-{\bf X}{\bf \hat{W}})^\top({\bf y}-{\bf X}{\bf \hat{W}})$

따라서 구하고 싶은것은 아래와 같음 

$$\hat{\bf W}^{LSE} = \underset{\bf \hat{W}}{\operatorname{argmin}} ~ loss(\hat{\bf W})$$

:::{.callout-warning}

아래의 수식

$$\hat{\bf W}^{LSE} = \underset{\bf \hat{W}}{\operatorname{argmin}} ~ loss(\hat{\bf W})$$

은 아래와 같이 표현해도 무방합니다. 

$$\hat{\bf W} = \underset{\bf W}{\operatorname{argmin}} ~ loss({\bf W})$$ 

마치 함수 $f(\hat{x})=({\hat x}-1)^2$ 을 $f(x)=(x-1)^2$ 이라고 표현할 수 있는 것 처럼요..
:::

`#`

`-` 더 나은 선으로 업데이트하기 위해서는 공식 "`더나은What = 원래What - 0.1*미분값`" 를 적용해야하고 이를 위해서는 `미분값`을 계산할 수 있어야 함. 

:::{.callout-important}

경사하강법을 좀 더 엄밀하게 써보자. 경사하강법은 $loss(\hat{\bf W})$를 최소로 만드는 $\hat{\bf W}$를 컴퓨터로 구하는 방법인데, 구체적으로는 아래와 같다. 

`1`. 임의의 점 $\hat{\bf W}$를 찍는다. 

`2`. 그 점에서 순간기울기를 구한다. 즉 $\left.\frac{\partial}{\partial {\bf W}}loss({\bf W})\right|_{{\bf W}=\hat{\bf W}}$ 를 계산한다. 

`3`. $\hat{\bf W}$에서의 순간기울기의 부호를 살펴보고 부호와 반대방향으로 움직인다. 이때 기울기의 절대값 크기와 비례하여 보폭(=움직이는 정도)을 각각 조절한다. 즉 아래의 수식에 따라 업데이트 한다. 

- $\hat{\bf W} \leftarrow \hat{\bf W} - \alpha \times \left.\frac{\partial}{\partial {\bf W}}loss({\bf W})\right|_{{\bf W}=\hat{\bf W}}$

여기에서 맨 마지막 수식을 간단하게 쓴 것이 `더나은What = 원래What - 0.1*미분값` 이다.
:::

`-` 미분값을 계산하는 방법1

````python
# 손실 8587.6875 를 계산하는 또 다른 방식
def l(w0,w1):
    yhat = w0 + w1*x
    return torch.sum((y-yhat)**2)
````

````python
l(-5,10)
````

````python
h=0.001
print((l(-5+h,10) - l(-5,10))/h)
print((l(-5,10+h) - l(-5,10))/h)
````

*일단 이거로 업데이트해볼까?*

````python
# 더나은What = 원래What - 0.1*미분값
# [-5,10] - 0.001 * [-1341.7968,1190.4297]
````

````python
sssss = What - 0.001 * torch.tensor([[-1341.7968],[1190.4297]])
sssss
````

````python
plt.plot(x,y,'o')
plt.plot(x,X@What,'-') # 원래What: 주황색
plt.plot(x,X@sssss,'-') # 더나은What: 초록색
````

- 잘 된 것 같긴한데..
- 미분구하는게 너무 어려워.. 
- 다른 방법 없을까? 

:::{.callout-important}

사실 이 방법은 

- $\frac{\partial}{\partial w_0}loss(w_0,w_1) \approx \frac{loss(w_0+h,w_1)-loss(w_0,w_1)}{h}$
- $\frac{\partial}{\partial w_1}loss(w_0,w_1) \approx \frac{loss(w_0,w_1+h)-loss(w_0,w_1)}{h}$

이 계산을 이용하여 

- $\frac{\partial}{\partial {\bf W}}loss({\bf W}):= \begin{bmatrix} \frac{\partial}{\partial w_0} \\ \frac{\partial}{\partial w_1}\end{bmatrix}loss({\bf W}) =  \begin{bmatrix} \frac{\partial}{\partial w_0}loss({\bf W}) \\ \frac{\partial}{\partial w_1}loss({\bf W})\end{bmatrix}  =  \begin{bmatrix} \frac{\partial}{\partial w_0}loss(w_0,w_1) \\ \frac{\partial}{\partial w_1}loss(w_0,w_1)\end{bmatrix}$

를 계산한 것이라 볼 수 있죠
:::

`-` 미분값을 계산하는 방법2 

````python
## 약간의 지식이 필요함. 
# loss = (y-XWhat)'(y-XWhat)
# = (y'-What'X')(y-XWhat)
# = y'y-y'XWhat -What'X'y + What'X'XWhat 
# loss를 What으로 미분
# loss' = -X'y - X'y + 2X'XWhat
````

````python
-2*X.T@y + 2*X.T@X@What
````

:::{.callout-important}

이 방법은 $loss({\bf W})$의 미분을 구할수 있어야 사용가능합니다. 즉 

$$\frac{\partial}{\partial {\bf W}}loss({\bf W})= -2{\bf X}^\top {\bf y} + 2{\bf X}^\top {\bf X}{\bf W}$$

를 계산할 수 있어야 합니다.
:::

`-` 미분값을 계산하는 방법3 -- 이 패턴을 외우세여

````python
What = torch.tensor([[-5.0],[10.0]],requires_grad=True)
What
````

````python
yhat = X@What
loss = torch.sum((y-yhat)**2)
loss
````

````python
loss.backward() # loss를 미분하라.. 꼬리표가 있게 한 What으로.. 
````

````python
What.grad
````

`#` 1회 업데이트 과정을 차근차근 시각화하며 정리해보자. 

````python
alpha = 0.001 
print(f"{What.data} -- 수정전")
print(f"{-alpha*What.grad} -- 수정하는폭")
print(f"{What.data-alpha*What.grad} -- 수정후")
print(f"{torch.tensor([[2.5],[4]])} -- 참값(이건 비밀~~)")
````

`#`

## C. 3단계 -- iteration (=learn = estimate $\bf{\hat W}$)

`-` 이제 1단계와 2단계를 반복만하면된다. 그래서 아래와 같은 코드를 작성하면 될 것 같은데... 

```Python
What = torch.tensor([[-5.0],[10.0]],requires_grad=True) # 최초의 직선을 만드는 값
for epoc in range(30):
    yhat = X@What 
    loss = torch.sum((y-yhat)**2)
    loss.backward()
    What.data = What.data - 0.001 * What.grad
```

돌려보면 잘 안된다. 

`-` 아래와 같이 해야한다. 

````python
What = torch.tensor([[-5.0],[10.0]],requires_grad=True) # 최초의 직선을 만드는 값
for epoc in range(30):
    yhat = X@What 
    loss = torch.sum((y-yhat)**2)
    loss.backward()
    What.data = What.data - 0.001 * What.grad
    What.grad = None 
````

````python
plt.plot(x,y,'o',label=r"observed: $(x_i,y_i)$")
plt.plot(x,X@What.data,'--o', label=r"estimated: $(x_i,\hat{y}_i)$ -- after 30 iterations (=epochs)", alpha=0.4 )
plt.legend()
````

`-` 왜? `loss.backward()` 는 아래의 역할을 하는것 처럼 이해되었지만 

- `What.grad` $\leftarrow$ `What에서미분값`

실제로는 아래의 역할을 수행하기 때문이다. (컴퓨터공학적인 이유로..)

- `What.grad` $\leftarrow$ `What.grad` + `What에서미분값`

:::{.callout-note}

`What.grad` $\leftarrow$ `What.grad` + `What에서미분값` 임을 확인하기 위해서.. 약간의 테스트를 했습니다. 

먼저 

```Python
What = torch.tensor([[-5.0],[10.0]],requires_grad=True) # 최초의 직선을 만드는 값
print(What.data)
print(What.grad)
```

를 확인한뒤 아래를 반복실행해봤을때 

```Python
yhat = X@What 
loss = torch.sum((y-yhat)**2)
loss.backward() # 
print(What.data)
print(What.grad)
```

`What.data`와 `What.grad` 값이 계속 일정하게 나온다면 

- `What.grad` $\leftarrow$ `What에서미분값`

이와 같은 계산이 진행되는 것이겠고, `What.grad`의 값이 자꾸 커진다면 

- `What.grad` $\leftarrow$ `What.grad` + `What에서미분값`

이와 같은 계산이 진행되는 것이겠죠? 
:::
