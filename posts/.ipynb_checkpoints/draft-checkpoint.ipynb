# 3. 기호

## A. 기호를 중시하는 사람

`-` 기호만이 우리를 구원할 수 있다. 

`-` 다른의미를 가진다면 다른기호를 써야 한다. 

`-` 하지만 현실은 그렇지 않음. 

- $x^2 +2x +1=0$ 에서 $x$를 의미하는 기호와
- $f(x)=x^2+2x+1$에서 $x$를 의미하는 기호는 같은가? 

`-` 되게 미묘한 차이 -- 아래는 같은 $x$를 의미하는게 아니다. 

`# 고등학교 문제` -- $f(x)=(x-2)^2$ 의 최소값을 구하자. 

(sol)

$f(x)=(x-2)^2$이고 $f''(x)=2>0$이다. 따라서 $f'(x)=2(x-2)=0$을 만족하는 $x$가 최소값이다. 따라서 $x=2$. 

`#`

`-` 그래서 산공에서는 최적값을 의미하는 기호로 보통 $\star$를 쓰기도 한다. 

`# 고등학교 문제` -- $f(x)=(x-2)^2$ 의 최소값을 구하자. 

(sol)

$f(x)=(x-2)^2$이고 $f''(x)=2>0$이다. 따라서 $2(x^\star-2)=0$을 만족하는 $x^\star$가 최소값이다. 따라서 $x^\star=2$. 

`#`

# 5. 회귀분석

## A. 문제

`-` 모형

$$y_i = w_0^\ast + w_1^\ast x_i + \epsilon_i$$

````python
torch.manual_seed(43052)
x = torch.randn(100).sort()[0]
eps = torch.randn(100)*0.5
w0ast = 2.5
w1ast = 4 
y = w0ast + w1ast*x + eps
````

`-` 시각화

````python
plt.plot(x,y,'o')
#plt.plot(x,4*x + 2.5,'--')
````

## B. 풀이1 -- 지난시간

(엄밀하게)

````python
def l(w0,w1):
    return torch.sum((y-(w0 + w1*x))**2)
````

````python
w0tilde = -5
w1tilde = 10
````

````python
alpha = 0.001
for epoch in range(30):    
    w0tilde_grad = (l(w0tilde+0.001,w1tilde) - l(w0tilde,w1tilde))/0.001
    w1tilde_grad = (l(w0tilde,w1tilde+0.001) - l(w0tilde,w1tilde))/0.001
    w0tilde = w0tilde - alpha * w0tilde_grad
    w1tilde = w1tilde - alpha * w1tilde_grad
````

````python
w0hat = w0tilde
w1hat = w1tilde
````

````python
plt.plot(x,y,'o')
plt.plot(x,w0hat+w1hat*x,'--')
````

(대충)

````python
def l(w0,w1):
    return torch.sum((y-(w0 + w1*x))**2)
````

````python
w0hat = -5
w1hat = 10
````

````python
for epoch in range(30):    
    w0hat_grad = (l(w0hat+0.001,w1hat) - l(w0hat,w1hat))/0.001
    w1hat_grad = (l(w0hat,w1hat+0.001) - l(w0hat,w1hat))/0.001
    w0hat = w0hat - 0.001 * w0hat_grad
    w1hat = w1hat - 0.001 * w1hat_grad
````

````python
plt.plot(x,y,'o')
plt.plot(x,w0hat+w1hat*x,'--')
````

````python
w0hat, w1hat
````

## C. 풀이2 -- 토치사용

````python
w0hat = torch.tensor(-5.00,requires_grad=True)
w1hat = torch.tensor(10.00,requires_grad=True)
````

````python
for epoch in range(30):    
    yhat = w0hat + w1hat*x 
    loss = torch.sum((y-yhat)**2)
    loss.backward()
    w0hat.data = w0hat.data - 0.001 * w0hat.grad
    w1hat.data = w1hat.data - 0.001 * w1hat.grad
    w0hat.grad = None
    w1hat.grad = None
````

````python
plt.plot(x,y,'o')
plt.plot(x,w0hat.data+w1hat.data*x,'--')
````

````python
w0hat, w1hat
````

## D. 풀이3 -- 벡터화

`-` 아이디어: 아래의 모형 

$$y_i = w_0^\ast + w_1^\ast x_i + \epsilon_i$$

을 아래와 같이 해석하자. 

$$
\begin{bmatrix}
y_1 \\
y_2 \\
\vdots \\
y_n
\end{bmatrix}
=
\begin{bmatrix}
1 & x_1 \\
1 & x_2 \\
\vdots & \vdots \\
1 & x_n
\end{bmatrix}
\begin{bmatrix}
w_0^\ast \\
w_1^\ast
\end{bmatrix}
+
\begin{bmatrix}
\epsilon_1 \\
\epsilon_2 \\
\vdots \\
\epsilon_n
\end{bmatrix}
$$

`-` 최초의 예측값은 

````python
w0hat = torch.tensor(-5.00,requires_grad=True)
w1hat = torch.tensor(10.00,requires_grad=True)
yhat = w0hat + w1hat * x
yhat[:5]
````

이런식이었는데, 이제는 아래와 같이 구할 수 있다. 

````python
X = torch.stack([torch.ones(100),x],axis=1)
Y = y.reshape(-1,1)
What = torch.tensor([[-5.00],[10.00]],requires_grad = True)
````

````python
Yhat = X@What
Yhat[:5]
````

`-` 첫미분은 원래 아래와 같이 구했는데.. 

````python
yhat = w0hat + w1hat*x 
loss = torch.sum((y-yhat)**2)
loss.backward()
w0hat.grad, w1hat.grad
````

이제 아래와 같이 하면 된다. 

````python
Yhat = X@What
loss = torch.sum((Y-Yhat)**2)
loss.backward()
What.grad
````

`-` 업데이트는 원래 아래와 같이 했는데 

````python
w0hat.data = w0hat.data - 0.001 * w0hat.grad
w1hat.data = w1hat.data - 0.001 * w1hat.grad
w0hat, w1hat
````

이제 아래와 같이 하면 된다. 

````python
What.data = What.data - 0.001 * What.grad 
What
````

`-` 그래서 새로운 코드를 정리하면 

````python
X = torch.stack([torch.ones(100),x],axis=1)
What = torch.tensor([[-5.00], [10.00]],requires_grad=True)
````

````python
for epoch in range(30):    
    Yhat = X@What
    loss = torch.sum((Y-Yhat)**2)
    loss.backward()
    What.data = What.data - 0.001*What.grad
    What.grad = None
````

````python
What
````

## E. 풀이4 -- MSE 

````python
X = torch.stack([torch.ones(100),x],axis=1)
What = torch.tensor([[-5.00],[10.00]],requires_grad=True)
````

````python
for epoch in range(30):    
    Yhat = X@What
    loss = torch.mean((Y-Yhat)**2)
    loss.backward()
    What.data = What.data - 0.1*What.grad
    What.grad = None
````

````python
What
````

## F. 풀이5 -- MSE함수 

````python
X = torch.stack([torch.ones(100),x],axis=1)
What = torch.tensor([[-5.00], [10.00]],requires_grad=True)
loss_fn = torch.nn.MSELoss()
````

````python
for epoch in range(30):    
    Yhat = X@What
    loss = loss_fn(Yhat,Y)
    loss.backward()
    What.data = What.data - 0.1*What.grad
    What.grad = None
````

`-` 이제 정리하면 

````python
X = torch.stack([torch.ones(100),x],axis=1)
net = torch.nn.Linear(in_features=2,out_features=1,bias=False)
net.weight.data = torch.tensor([[-5.00], [10.00]]).T
loss_fn = torch.nn.MSELoss()
````

````python
for epoch in range(30):    
    Yhat = net(X)
    loss = loss_fn(Yhat,Y)
    loss.backward()
    net.weight.data = net.weight.data - 0.1*net.weight.grad
    net.weight.grad = None
````

## H. 풀이7 -- `optimizer`

`-` 옵티마이저를 쓰면 업데이트 수식을 외울 필요가 없다. 

````python
X = torch.stack([torch.ones(100),x],axis=1)
net = torch.nn.Linear(in_features=2,out_features=1,bias=False)
net.weight.data = torch.tensor([[-5.00], [10.00]]).T
loss_fn = torch.nn.MSELoss()
optimizer = torch.optim.SGD(net.parameters(),lr=0.1)
````

````python
for epoch in range(30):    
    Yhat = net(X)
    loss = loss_fn(Yhat,Y)
    loss.backward()
    optimizer.step()
    net.weight.grad = None
````

## I. 풀이8 -- `optimizer`, `zero_grad`

````python
X = torch.stack([torch.ones(100),x],axis=1)
net = torch.nn.Linear(in_features=2,out_features=1,bias=False)
net.weight.data = torch.tensor([[-5.00], [10.00]]).T
loss_fn = torch.nn.MSELoss()
optimizer = torch.optim.SGD(net.parameters(),lr=0.1)
````

## J. 풀이9 -- `net`, `bias`

`-` `net`에서 `bias=False`가 의미하는건 뭘까? 

`-` 간단함.

**`bias= False`**

```Python
net(inp) = inp @ net.weight.T 
```

**`bias= True`**

```Python
net(inp) = inp @ net.weight.T + net.bias
```


````python
X = torch.stack([torch.ones(100),x],axis=1)
net = torch.nn.Linear(in_features=1,out_features=1,bias=True)
net.weight.data = torch.tensor([[10.00]]).T
net.bias.data = torch.tensor([-5.00])
loss_fn = torch.nn.MSELoss()
optimizer = torch.optim.SGD(net.parameters(),lr=0.1)
````

````python
for epoch in range(30):    
    Yhat = net(X[:,[1]])
    loss = loss_fn(Yhat,Y)
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
````
